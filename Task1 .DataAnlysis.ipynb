```python
Q1 = merged_df["tot_sales"].quantile(0.25)
Q3 = merged_df['tot_sales'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = merged_df[(merged_df['tot_sales'] < lower_bound) | (merged_df['tot_sales'] > upper_bound)]
```

```python
from scipy.stats import zscore

merged_df['z_score'] = zscore(merged_df['tot_sales'])
outliers = merged_df[(merged_df['z_score'] > 3) | (merged_df['z_score'] < -3)]
```
